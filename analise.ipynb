```python
import pandas as pd
```

```python

df = pd.read_csv("dados_s3_sinteticos.csv")
df.describe()
```

```python
custo_por_bucket = df.groupby("bucket")["custo_mensal_usd"].sum().sort_values(ascending=False)
custo_por_bucket
```

```python
custo_por_classe = df.groupby("storage_class")["custo_mensal_usd"].sum().sort_values(ascending=False)
custo_por_classe
```

```python
custo_bucket_classe = df.groupby(["bucket", "storage_class"])["custo_mensal_usd"].sum().sort_values(ascending=False)
custo_bucket_classe
```

```python

custo_por_region = df.groupby(["bucket","regiao"],)["custo_mensal_usd"].sum().sort_values(ascending=False)
custo_por_region
```
